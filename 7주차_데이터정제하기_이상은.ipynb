```python
import pandas as pd # 판다스를 불러올거야, pd라고 지정할거야
ns_df = pd.read_csv('ns_202104.csv', low_memory=False) #판다스가 csv 읽어봐
ns_df.head() #처음 다섯 개 행을 출력 ㄱㄱ
```

```python
ns_book = ns_df.loc[:, '번호':'등록일자'] #[ ] 번호열부터 등록일자 열까지 전체 행을 선택
ns_book.head()
```

```python
print(ns_df.columns) #데이터 프레임 열 이름이 저장된 콜럼 속성 확인
```

```python
print(ns_df.columns[0]) #콜럼 속성은 판다스의 index 클래스 객체
```

```python
#원소별 비교를 활용하여 ns_df.columns에서 'Unnamed: 13' 열이 아닌 것을 표시하는 배열 만들기
# "!" 는 비교연산자!!!
#columns 속성과 비교하여 'Unnamed: 13'열이 아닌 것은 True가 되고 'Unnamed: 13'열은 Flase
#반환된 결과는 Numpy Array

#교수님이 오늘 이것만 알고 가면 된다 하심 시험에 비슷하게 나온대
ns_df.columns != 'Unnamed: 13'
```

```python
#원소별 비교 연산으로 얻은 불리언 배열을 selected_columns 변수에 저장하고 판다스 데이터 프레임의 loc 메서드에 전달하여 True인 열의 행만 선택
selected_columns = ns_df.columns != 'Unnamed: 13'
ns_book = ns_df.loc[:, selected_columns] # [ ] True인 열의 모든 행을 선택
ns_book.head()
```

```python
#데이터프레임 중간에 있는 '부가기호' 열을 제외시키기
selected_columns = ns_df.columns != '부가기호'
ns_book = ns_df.loc[:, selected_columns]
ns_book.head()
```

```python
#drop() 메서드
#데이터 프레임의 행이나 열을 삭제하는 !!!
#첫 번째 매개변수에 삭제하려는 열 이름을 전달하고 axis 매개변수를 1로 지정
#ns_df 데이터 프레임에서 불필요한 'Unnamed: 13'열을 삭제하기

ns_book = ns_df.drop('Unnamed: 13', axis=1) #삭제하려는 열 이름, axis 매개변수에 1을 지정하면 열을 삭제
ns_book.head()
```

```python
#drop() 메서드로 중간에 있는 열을 간단하게 제외
#또한 첫 뻔재 매개변수에 제외할 열 이름을 리스트 형식으로 여러 개 지정
#'부가기호'열과 'Unnamed: 13'열 제외하기

ns_book = ns_df.drop(['부가기호', 'Unnamed: 13'], axis=1) #[ ] 여러 개 열을 삭제하려면 리스트 형식을 사용
ns_book.head()
```

```python
#drop() 메서드에 inplace 매개변수를 True로 지정하면 현재 선택한 데이터 프레임을 바로 수정
#앞서 생성한 ns_book 데이터프레임의 '주제분류번호' 열 삭제하기

ns_book.drop('주제분류번호', axis=1, inplace=True) # 선택한 데이터 프레임에 덮어씁니다
ns_book.head()
```

```python
#dropna() 메서드
#기본적으로 NaN이 하나 이상 포함된 행이나 열을 삭제
#axis 매개변수를 1로 지정하여 ns_df 데이터프레임에서 NaN이 포함된 열을 삭제하기
ns_book = ns_df.dropna(axis=1)
ns_book.head()
```

```python
#모든 값이 NaN인 열을 삭제하려면 dropna() 메서드에 how 매개변수를 'all'로 지정
ns_book = ns_df.dropna(axis=1, how='all')
ns_book.head()
```

```python
#행 삭제하기
#행을 삭제할 때도 drop() 메서드 사용
#axis 매개변수를 0으로 지정하면 행을 삭제할 수 있지만, 기본값이 0이므로 생략해도 상관 ㄴㄴ

#처음 2개 행 삭제하기!

ns_book2 = ns_book.drop([0, 1]) #인덱스 0부터 1까지 2개 행을 선택합니다
ns_book2.head()
```

```python
#[ ] 연산자와 슬라이싱
#[ ] 연산자에 슬라이싱이나 불리언 배열을 전달하면 행을 선택
# 인덱스가 0, 1인 행을 제외한 모든 행을 선택하기 위해 다음처럼 [] 연산자에 슬라이싱을 사용

ns_book2 = ns_book[2:]
ns_book2.head()
```

```python
#[ ] 연산자에 슬라이싱을 사용하면 loc 메서드에 슬라이싱을 사용하는 것과는 다르게 파이썬의 슬라이싱처럼 마지막 인덱스를 포함하지 않음
#[0:2]로 작성하여 처음 두 개의 행만 선택

ns_book2 = ns_book[0:2] # 인덱스 0과 1을 선택합니다. 2는 포함하지 않습니다.
ns_book2.head()
```

```python
#[ ] 연산자와 불리언 배열
# 불리언 배열을 사용해서 행 선택
# 비교 연산자를 활용해 원하는 행은 True로 표시하고 제외할 행은 False로 표시한 불리언 배열을 만들어 사용
# 출판사가 '한빛미디어'인 행만 선택

# 이것도 시험에 내셨나봄;;

seleced_rows = ns_df['출판사'] == '한빛미디어'
ns_book2 = ns_df[seleced_rows]
ns_book2.head()
```

```python
#불리언 배열을 만드는 조건은 [] 연산자에 바로 넣어 사용
#대출건수가 1000 이하 행 모두 삭제? - 반대로 ns_book['대출건수'] > 1000과 같이 조건 넣기 -> 대출건수 1000 넘는 행 선택

ns_book2 = ns_df[ns_df['대출건수'] > 1000]
ns_book2.head()
```

중복된 행 찾기

중복된 행은 duplicated() 메서드를 사용하여 검사

- 중복된 행 중에서 처음 행을 제외한 나머지 행을 True로, 그 외에 주복되지 않은 나머지 모든 행은 False로 표시한 불리언 배열을 반환

```python
#sum() 함수를 함께 사용하면 True를 1로 인식하기 때문에 duplicated() 메서드로 만든 불리언 배열에서 중복된 행의 개수를 셀 수 있음
#불필요한 열을 정리한 ns_book 데이터프레임에 중복된 행이 있는지 확인

sum(ns_book.duplicated())
```

```python
#도서명, 저자, ISBN을 기준으로 중복된 행이 있는지 찾아보기
# - 일부 열을 기준으로 중복된 행을 찾으려면 dubplicated() 메서드의 subset 매개변수에 기준열을 나열

sum(ns_book.duplicated(subset=['도서명', '저자', 'ISBN']))
```

```python
from os import dup
#도서명, 저자, ISBN을 기준으로 어떤 데이터가 중복되었는지 확인
# duplicated() 메서드에 keep 매개변수를 False로 지정하여 중복된 모든 행을 True로 표시
# 이 데이터를 ns_book3 변수에 저장하여 확인

dup_rows = ns_book.duplicated(subset=['도서명', '저자', 'ISBN'], keep=False) # keep=False) : 중복된 행을 모두 True로 표시한 불리언 배열을 반환합니다
ns_book3 = ns_book[dup_rows]
ns_book3.head()
```

```python
#그룹별로 모으기
# 같은 도서의 대출건수를 하나로 합ㅂ치기
# groupby() 메서드의 by 매개변수에는 행을 합칠 때 기준이 되는 열을 지정
# 도서명, 출판사, 발행연도, 대출건수 열을 기준으로 행을 합치기

# ns_book 데이터프레임의 전체 열을 사용하는 대신, 그룹으로 묶을 기준 열과 '대출건수' 열만 선택하여 합치기
count_df = ns_book[['도서명', '저자', 'ISBN', '권', '대출건수']] # 그룹으로 묶을 기준 열
```

```python
#결과를 반환한 count_df 데이터 프레임에 groupby() 메서드를 적용
# 정수 타입인 '대출건수' 열은 sum() 메서드를 사용
# NaN이 포함되어 있는 행을 삭제하고 계산하면 대출 건수 합계에서 빠지기 때문에 이를 막기 위해 dropna 매개변수를 False로 지정
group_df = count_df.groupby(by=['도서명', '저자', 'ISBN', '권'], dropna=False) # NaN이 있는 행을 삭제하지 않습니다.
loan_count = group_df.sum()
```

```python
#또는 위와 같이 groupby()메서드와 sum()메서드를 나누어서 호출해도 되지만 판다스 사용자들은 두 메서드를 연이어 호출하는 것을 선호
#어떤 작업을 수행하는지 명확하게 드러낼 수 있기 때문임!!

loan_count = count_df.groupby(by=['도서명', '저자', 'ISBN', '권'],
                              dropna=False).sum()
loan_count.head()
```

head() 메서드로 데이터프레임을 출력하면 기준인 인덱스 열이 굵은 글씨로 표시됩니다. loan_count 데이터 프레임은 '도서명', '저자', 'ISBN', '권'까지 네 개의 열이 인덱스이고 각 책의 대출건수를 더한 결과가 '대출건수' 열에 저장된 것을 알 수 있습니다.

원본 데이터 업데이트 하기

원본 데이터프레임에 중복된 데이터 있음
so..
1. duplicated() 메서드로 중복된 행을 True로 표시한 불리언 배열을 만들기
2. 1번에서 구한 불리언 배열을 반전시켜서 중복되지 않은 고유한 행을 True로 표시
3. 2번에서 구한 불리언 배열을 사용해 원본 배열에서 고유한 행만 선택

이 과정으로 코드를 살펴봐야함!!!

```python
#중복된 행을 True로 표시한 불리언 배열을 반전시킬 때는 판다스의 "~연산자"를 사용!!
# -> 그 다음에 원본 배열에서 고유한 배열을 선택하여 copy() 메서드로 ns_book3 데이터 프레임을 제작
dup_rows = ns_book.duplicated(subset=['도서명', '저자', 'ISBN', '권']) #중복된 행을 True로 표시

# 하다가 말았다잉 여기부터 다시 ㄱㄱㄱ

unique_rows = ~dup_rows #불리언 배열을 반전시켜 고유한 행을 True로 표시
ns_book3 = ns_book[unique_rows].copy() #고유한 행만 선택
```

**copy() 메서드**
데이터 프레임의 복사본 만든다!

unique_rows 배열로 일부 행을 선택 후 제작한 ns_book3의 '대출건수' 열을 업데이트 해야함

판다스는 copy() 메서드를 사용하지 않으면 ns_book3 데이터프레임이 별도의 메모리 공간에 저장되는지 보장하지 않는다.

그래서 판다스에서 데이터를 업데이트 할 때는 항상 복사하는 게 좋다.

```python
#ns_book3에 중복된 행이 없는지 duplicated() 메서드로 확인
sum(ns_book3.duplicated(subset=['도서명', '저자', 'ISBN', '권']))
```

**원본 데이터 프레임 인덱스 설정**

ns_book3의 인덱스를 loan_count 데이터프레임의 인덱스와 동일하게 만들기!!

지정한 열을 인덱스로 설정할 땐 **set_index()** 메서드 사용

**inplace 매개변수**를 True로 지정 -> 새로운 데이터프레임 반환 X -> ns_book3 데이터프레임 수정

```python
#인덱스가 변경된 ns_book3 데이터프레임 첫 다섯 개 행 출력

ns_book3.set_index(['도서명', '저자', 'ISBN', '권'], inplace=True)
ns_book3.head()
```

```python
#ns_book3 데이터프레임의 '대출건수' 열을 loan_count 데이터프레임의 '대출건수' 열로 업데이트
#update() 메서드 사용

ns_book3.update(loan_count)
ns_book3.head()
```

```python
#업데이트 제대로 되었으면 인덱스 열 해제!
#reset_index() 메서드 활용
#set_index : 열을 인덱스로 설정

ns_book4 = ns_book3.reset_index()
ns_book4.head()
```

```python
#대출 건수 잘 합쳐졌는지 간단히 확인
#원본 데이터프레임 ns_book에 대출건수 100회 이상인 책 개수 세어보기

sum(ns_book['대출건수']>100)
```

```python
#새로 만든 ns_book4 데이터 프레임에서 100회 이상인 책의 개수 세어보기

sum(ns_book4['대출건수']>100)
```

대출 건수 100회 이상인 책이 훨씬 늘어남,,

**중복된 도서의 대출 건수가 합쳐짐**

그러나 여러 개의 열을 사용해 데이터프레임의 인덱스를 만들었다가 다시 해제했기 때문에 초기 데이터프레임의 열과 ns_book4 데이터프레임의 열 순서가 달라짐. 데이터 분석을 수행할 때 열의 순서가 문제가 되지는 않지만, 앞으로 작업을 진행할 때 혼동하지 않도록 **원래 열의 순서대로 맞추**자

열의 순서를 바꾸는 가장 간단한 방법!!

**[]연산자**에 원하는 열 이름을 **순서대로** 전달하는 것

```python
ns_book4 = ns_book4[ns_book.columns]
ns_book4.head()
```

```python
#지금까지 불필요한 데이터를 제거하여 만든 ns_book4 데이터 프레임을 다음 작업을 위해 저장

ns_book4.to_csv('ns_book4.csv', index=False)
```

```python
#중복된 행 찾는 방법

#중복된 행의 대출 건수 합치기 : groupby() 메서드와 sum() 메서드 사용
#대출 건수를 원본 데이터프레임에 업데이트 하기 위해 update() 메서드 사용
#이때 index 기준으로 업데이트 하기 위해 set_index()와 reset_index() 사용

#일괄 처리 함수

#하 걍 텍스트로 할 걸;; 길어지네,, 일욜인데 놀러가고싶다

#불필요한 행과 열을 제거하기 위해 작성했던 코드를 새로운 데이터에 적용하기 쉽도록 일괄 처리하는 data_cleaning() 함수를 만들어보자

def data_cleaning(filename) :
  """
  남산 도서관 장서 CSV 데이터 전처리 함수
  :param filename : CSV 파일 이름
  """
  # 파일을 데이터 프레임으로 읽습니다
  ns_df = pd.read_csv(filename, low_memory=False)
  #NaN인 열 삭제
  ns_book = ns_df.dropna(axis=1, how='all')
  #대출 건수를 합치기 위해 필요한 행만 추출하여 count_df 데이터 프레임을 만듭니다.
  count_df = ns_book[['도서명', '저자', 'ISBN', '권', '대출건수']]
  #도서명, 저자, ISBN, 권을 기준으로 대출건수를 groupby 합니다
  loan_count = count_df.groupby(by=['도서명', '저자', 'ISBN', '권'],
                                dropna=False).sum()
  #원본 데이터프레임에서 중복된 행을 제외하고 고유한 행만 추출하여 복사
  dup_rows = ns_book.duplicated(subset=['도서명', '저자', 'ISBN', '권'])
  unique_rows = ~dup_rows
  ns_book3 = ns_book[unique_rows].copy()
  #도서명, 저자, ISBN, 권을 인덱스로 설정합니다
  ns_book3.set_index(['도서명', '저자', 'ISBN', '권'], inplace=True)
  #loan_count에 저장된 누적 대출 건수를 업데이트 합니다
  ns_book3.update(loan_count)
  #인덱스를 재설정합니다
  ns_book4 = ns_book3.reset_index()
  #원본 데이터 프레임의 열 순서로 변경합니다
  ns_book4 = ns_book4[ns_book.columns]
  return ns_book4
```

```python
#제대로 동작하는지 보자잉
#다른 데이터 프레임을 비교할 때는 equals() 메서드를 사용한다

new_ns_book4 = data_cleaning('ns_202104.csv') # 원본 데이터를 전달하여 / 새로운 데이터 프레임을 만들고 / 이전에 만든 것과 동일한지 비교
ns_book4.equals(new_ns_book4)

#직독직해를 해보자면,, 새 북포는 원본데이터로 프레임 만들고 / 북포와 새북포가 같은지 봐라~
```
